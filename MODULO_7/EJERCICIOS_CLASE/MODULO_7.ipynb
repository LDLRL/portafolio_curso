# **Informe: Caso Aerolínea y las 5V’s de Big Data**

##  Introducción
Este informe analiza cómo se manifiestan las **5V’s del Big Data** (Volumen, Velocidad, Variedad, Veracidad y Valor) en el contexto de una **aerolínea internacional**.  
El objetivo es identificar ejemplos reales en este sector, explicar su importancia y proponer herramientas del ecosistema Big Data que ayuden a gestionarlas de forma eficiente.

---

##  1. Volumen
**Definición**: Se refiere a la **gran cantidad de datos** que se generan diariamente.  

**Ejemplo en la aerolínea**:  
- Millones de registros de pasajeros y reservas.  
- Historial de vuelos y transacciones.  
- Datos de mantenimiento de aviones y sensores IoT.  

**Herramientas recomendadas**:  
- **HDFS / Hadoop** → sistema distribuido para guardar enormes volúmenes de datos.  
- **Amazon S3 / Google BigQuery** → almacenamiento en la nube escalable.  

**Justificación**:  
Estas herramientas permiten manejar datos que **no caben en un solo servidor**, distribuyéndolos en clústeres o la nube, asegurando capacidad y disponibilidad a gran escala.  

---

##  2. Velocidad
**Definición**: Capacidad de procesar datos **en tiempo real** o casi en vivo.  

**Ejemplo en la aerolínea**:  
- Compras de boletos en línea que deben confirmarse al instante.  
- Sensores de aviones enviando datos durante el vuelo.  
- Check-in y movimiento de equipajes en aeropuertos.  

**Herramientas recomendadas**:  
- **Apache Kafka** → ingesta de datos en tiempo real desde múltiples fuentes.  
- **Apache Flink / Spark Streaming** → procesamiento continuo o en micro-lotes.  
- **AWS Kinesis** → alternativa en la nube con escalado automático.  

**Justificación**:  
Estas herramientas permiten **recibir y analizar datos en segundos o milisegundos**, fundamentales para detectar fraudes, ajustar precios dinámicos o dar alertas inmediatas.  

---

##  3. Variedad
**Definición**: Se refiere a los **diferentes tipos y formatos de datos**.  

**Ejemplo en la aerolínea**:  
- Datos estructurados: tablas de reservas en SQL.  
- Datos semiestructurados: JSON con información de pasajeros y vuelos.  
- Datos no estructurados: reseñas en redes sociales, fotos de equipajes, audios de reclamos.  

**Herramientas recomendadas**:  
- **MongoDB / Cassandra (NoSQL)** → para datos semiestructurados y flexibles.  
- **ElasticSearch** → para búsqueda y análisis de texto libre.  
- **Data Lake (S3, Azure Data Lake, HDFS)** → para almacenar todo tipo de formatos.  
- **Apache NiFi / Talend** → para integrar y transformar datos de múltiples fuentes.  

**Justificación**:  
Estas tecnologías permiten manejar datos **estructurados, semiestructurados y no estructurados** en un solo ecosistema, dando flexibilidad y evitando perder información valiosa.  

---

##  4. Veracidad
**Definición**: Asegurar que los datos sean **confiables, coherentes y sin errores**.  

**Ejemplo en la aerolínea**:  
- Un pasajero aparece como “pagado” en un sistema, pero “no pagado” en otro.  
- Información duplicada en reservas o registros incompletos de pasajeros.  
- Riesgos de fraude en pagos con tarjeta.  

**Herramientas recomendadas**:  
- **Apache Atlas** → gobernanza de datos, catálogo y linaje.  
- **ETL / Talend** → limpieza y validación de datos antes de analizarlos.  

**Justificación**:  
Apache Atlas permite **trazar el recorrido del dato (lineage)**, quién lo usó y con qué reglas, mientras ETL asegura que los datos estén **limpios y consistentes** antes de entrar a los sistemas de análisis.  

---

##  5. Valor
**Definición**: Convertir los datos en **decisiones y beneficios reales** para el negocio.  

**Ejemplo en la aerolínea**:  
- Predecir retrasos combinando datos históricos y clima.  
- Ajustar precios dinámicos según la demanda en tiempo real.  
- Detectar clientes leales y personalizar ofertas.  
- Optimizar rutas de vuelo para reducir costos.  

**Herramientas recomendadas**:  
- **Apache Spark (Spark SQL, MLlib, Streaming)** → análisis masivo, predicciones y aprendizaje automático.  
- **Power BI / Tableau** → visualización clara de KPIs y reportes.  
- **TensorFlow** → modelos avanzados de predicción y clasificación.  

**Justificación**:  
Spark convierte grandes volúmenes de datos en **análisis predictivo y decisiones rápidas**, mientras las herramientas de BI y ML permiten que los datos se traduzcan en **valor económico y estratégico**.  

---

## Tabla Resumen de las 5V’s

| V | Necesidad en la aerolínea | Herramientas recomendadas | Justificación |
|---|----------------------------|----------------------------|---------------|
| **Volumen** | Millones de reservas, pasajeros, vuelos, sensores | HDFS, S3, BigQuery | Escalar almacenamiento y procesamiento de datos masivos. |
| **Velocidad** | Transacciones en línea, sensores IoT, check-in en vivo | Kafka, Flink, Spark Streaming, Kinesis | Procesar datos en tiempo real para decisiones inmediatas. |
| **Variedad** | Tablas SQL, JSON, logs, redes sociales, imágenes | MongoDB, Cassandra, ElasticSearch, Data Lake, NiFi | Integrar múltiples formatos y fuentes en un solo ecosistema. |
| **Veracidad** | Evitar errores, duplicados, fraudes, inconsistencias | Apache Atlas, Talend (ETL) | Asegurar datos confiables y trazables. |
| **Valor** | Precios dinámicos, predicción de retrasos, personalización | Spark (SQL, MLlib), TensorFlow, Power BI | Convertir datos en decisiones y beneficios estratégicos. |

---

##  Conclusión
El caso de la aerolínea muestra cómo las **5V’s de Big Data** se manifiestan en la práctica y requieren un conjunto de herramientas específicas.  
- El **Volumen** se gestiona con sistemas distribuidos de almacenamiento.  
- La **Velocidad** con plataformas de streaming que procesan datos en segundos.  
- La **Variedad** con bases NoSQL y Data Lakes que soportan múltiples formatos.  
- La **Veracidad** con catálogos y gobernanza para asegurar confianza.  
- Y el **Valor** con Spark, ML y BI, que transforman datos en **ventaja competitiva real**.

## 1. Caso de uso elegido: Análisis de logs de un servidor web
Una empresa quiere procesar millones de registros de logs diarios (en formato texto/JSON) para:

## 2. Componentes de Spark que intervienen

- **Driver:**  
  Coordina el trabajo, planifica el flujo (DAG) y reparte las tareas.

- **Cluster Manager (ej. YARN, Kubernetes o Standalone):**  
  Decide en qué nodo correrá cada parte del trabajo.

- **Executors:**  
  Son los procesos en los nodos que realmente ejecutan las operaciones sobre los datos.

- **RDDs:**  
  Representan los registros distribuidos de los logs cuando se leen desde los archivos.

- **DataFrames:**  
  Los logs se convierten a DataFrame para analizarlos con SQL (ej: agrupar por hora, contar errores).

- **Spark SQL:**  
  Permite ejecutar consultas tipo SQL sobre los logs para responder preguntas de negocio.

## 4. Justificación: ¿Por qué Spark y no otra herramienta?

- **Vs Hadoop MapReduce:**  
  Spark es mucho más rápido gracias al procesamiento en memoria. Analizar millones de registros puede tomar segundos o minutos en vez de horas.  

- **Vs Flink**: Flink es más potente en *streaming puro* (evento por evento), pero en este caso los logs se procesan en **lotes diarios (batch)**, lo que Spark maneja de manera excelente

- **Vs Apache Storm**: Storm está especializado en *streaming en tiempo real*, pero carece de la versatilidad de Spark, que combina batch, streaming, SQL y machine learning en una sola plataforma.  

- **Vs Hive**: Hive está pensado para consultas SQL sobre Hadoop, pero suele ser más lento (minutos u horas). Spark SQL aprovecha la memoria para ejecutar consultas mucho más rápido.  

- **Vs Kafka (Streams)**: Kafka es excelente para la **ingesta de datos en tiempo real**, pero no tiene capacidades de análisis avanzado. Spark Structured Streaming puede procesar y analizar los mismos datos con mayor flexibilidad.  

- **Vs BigQuery / Snowflake**: estos Data Warehouses en la nube son muy buenos para consultas SQL sobre datos estructurados, pero Spark ofrece más flexibilidad al trabajar también con datos semiestructurados y no estructurados, además de integrar machine learning.

# Comparativa enriquecida: Apache Spark vs Hadoop MapReduce vs Apache Flink

## Tabla comparativa

| **Criterio**                 | **Apache Spark**                                                                                                                   | **Hadoop MapReduce**                                                                                  | **Apache Flink**                                                                                                         |
|-------------------------------|------------------------------------------------------------------------------------------------------------------------------------|-------------------------------------------------------------------------------------------------------|--------------------------------------------------------------------------------------------------------------------------|
| **Modelo de procesamiento**  | Procesamiento por lotes y streaming (con Spark Streaming y Structured Streaming, basado en micro-batches).                         | Procesamiento por lotes únicamente. Cada job genera tareas que leen desde disco y escriben resultados en disco. | Procesamiento en tiempo real (streaming nativo, evento por evento) y también soporta batch.                             |
| **Uso de memoria**            | Computación en memoria (**in-memory**) → muy rápido, ideal para operaciones iterativas (ej. ML), pero exige más RAM.               | Computación basada en disco → más lento, pero consume menos memoria y puede manejar volúmenes muy grandes.       | Computación en memoria optimizada para flujos continuos y tolerancia a fallos con checkpoints y snapshots.               |
| **Lenguajes soportados**      | Scala, Java, Python (PySpark), R.                                                                                                 | Principalmente Java (algunos wrappers para otros lenguajes, pero más limitados).                               | Scala, Java, Python (API más madura en Scala/Java).                                                                      |
| **Latencia**                  | Baja latencia en streaming (segundos o subsegundos), aunque mayor que Flink, por usar micro-batches.                              | Alta latencia: minutos u horas (cada lote debe leerse y escribirse en disco).                                   | Muy baja latencia (milisegundos), diseñado para *event-time processing* (procesar eventos según su timestamp real).       |
| **Casos de uso recomendados** | - Análisis de grandes volúmenes de datos. <br>- ETL (extracción, transformación, carga). <br>- Machine Learning con MLlib. <br>- Análisis batch y streaming unificado. | - Procesamiento histórico de logs, transacciones o datos masivos. <br>- Escenarios donde la prioridad es costo y robustez, no la velocidad. | - Procesamiento en tiempo real: IoT, fraudes financieros, telecomunicaciones, sistemas de monitoreo y alertas inmediatas. |
| **Facilidad de uso**          | APIs de alto nivel (DataFrames, SQL, MLlib). Ecosistema maduro y gran comunidad. Relativamente fácil para analistas de datos.     | Difícil de programar: requiere mucho código en Java y conocimientos técnicos de Hadoop.                           | APIs amigables (DataStream API), aunque menos maduro que Spark. Requiere más experiencia para arquitecturas complejas.   |
| **Ejemplos de uso real**      | Netflix (recomendaciones personalizadas), Uber (predicción de demanda), Alibaba (análisis de ventas).                             | Yahoo (procesamiento de correos y búsquedas), Facebook en sus primeros sistemas de analítica masiva.             | ING Bank (detección de fraudes en tiempo real), Ericsson (procesamiento de datos IoT en telecomunicaciones).             |

---

## Escenarios recomendados

1. **Análisis en tiempo real de redes sociales** → **Apache Flink**  
   - Diseñado para *streaming* puro, con muy baja latencia.  
   - Ejemplo: monitorear menciones en Twitter para lanzar alertas en milisegundos.  

   *(Alternativa: Spark Streaming si ya existe infraestructura Spark, aunque con micro-batches más lentos que Flink).*  

---

2. **Procesamiento por lotes de logs históricos** → **Hadoop MapReduce** o **Apache Spark**  
   - **Hadoop**: más barato para procesar grandes lotes cuando la latencia no es crítica.  
   - **Spark**: más rápido gracias a in-memory y más flexible con APIs SQL/ML.  

   Ejemplo: analizar logs de servidores de un mes completo para detectar fallos recurrentes.  

---

3. **Modelo de aprendizaje automático iterativo** → **Apache Spark**  
   - Spark MLlib está optimizado para algoritmos iterativos (ej. regresión, clustering, árboles).  
   - Hadoop no es eficiente para ML porque escribe en disco tras cada iteración.  
   - Flink es fuerte en streaming, pero menos usado en ML.  

   Ejemplo: entrenar un modelo de predicción de demanda sobre millones de registros históricos.  

---

## Conclusión

- **Spark**: versátil, rápido y con ecosistema amplio (batch + streaming + ML). Es la opción preferida en la mayoría de proyectos modernos de Big Data.  
- **Hadoop MapReduce**: sigue siendo útil en escenarios de procesamiento **masivo por lotes** con bajo costo, donde la latencia no importa.  
- **Flink**: insuperable para **tiempo real** con latencias muy bajas, en sectores como banca, IoT y telecomunicaciones.

Explora y transforma datos con RDDs

Contexto:

Una de las claves para trabajar con datos distribuidos en Spark es conocer cómo manipular RDDs mediante transformaciones. Esta actividad te permitirá practicar con datos reales y aplicar operaciones fundamentales.

Consigna:

A partir de un conjunto de datos (puede ser un archivo .txt simulado con texto libre o una lista), aplica transformaciones utilizando métodos como filter, map, flatMap, distinct y sortBy para limpiar y reorganizar la información.


Paso a paso:

1. Carga el archivo o lista usando sc.textFile() o sc.parallelize().

2. Utiliza flatMap() para separar el texto en palabras.

3. Aplica filter() para eliminar palabras vacías o con menos de 4 letras.

4. Usa map() para crear pares (palabra, 1).

5. Aplica reduceByKey() para contar la frecuencia de cada palabra.

6. Ordena los resultados con sortBy() y mostra las 5 más. frecuentes.

7. Muestra el resultado con take(5) o collect().

````python
# Paso 0: Configurar Apache Spark en Google Colab
!apt-get install openjdk-8-jdk-headless -qq > /dev/null
!wget -q https://archive.apache.org/dist/spark/spark-3.4.0/spark-3.4.0-bin-hadoop3.tgz
!tar xf spark-3.4.0-bin-hadoop3.tgz
!pip install -q findspark

import os
os.environ["JAVA_HOME"] = "/usr/lib/jvm/java-8-openjdk-amd64"
os.environ["SPARK_HOME"] = "/content/spark-3.4.0-bin-hadoop3"

import findspark
findspark.init()

from pyspark import SparkContext
sc = SparkContext.getOrCreate()

# Paso 1: Carga la lista de texto simulando un archivo
data = [
    "Explora y transforma datos con RDDs",
    "Una de las claves para trabajar con datos distribuidos en Spark",
    "conocer cómo manipular RDDs mediante transformaciones.",
    "Esta actividad te permitirá practicar con datos reales",
    "y aplicar operaciones fundamentales."
]
rdd = sc.parallelize(data)

# Paso 2: Utiliza flatMap para separar el texto en palabras
words = rdd.flatMap(lambda line: line.split(" "))

# Paso 3: Filtra palabras vacías o con menos de 4 letras
filtered_words = words.filter(lambda w: w.strip() != "" and len(w) >= 4)

# Paso 4: Usa map para crear pares (palabra, 1)
pairs = filtered_words.map(lambda w: (w.lower(), 1))

# Paso 5: ReduceByKey para contar la frecuencia de cada palabra
counts = pairs.reduceByKey(lambda a, b: a + b)

# Paso 6: Ordena los resultados con sortBy para mostrar las más frecuentes
sorted_counts = counts.sortBy(lambda x: x[1], ascending=False)

# Paso 7: Muestra las 5 palabras más frecuentes
top5 = sorted_counts.take(5)

print("Las 5 palabras más frecuentes con 4 o más letras son:")
for palabra, frecuencia in top5:
    print(f"{palabra}: {frecuencia}")
````

2) Normaliza nombres de columnas y castea numéricos

Muchos CSV traen acentos o espacios. Los normalizamos a nombres seguros (snake_case) y casteamos métricas a numérico para poder agrupar y sumar sin errores.

````python
import unicodedata
import re
from pyspark.sql import functions as F

# Función para normalizar nombres de columnas
def normalize_col(s):
    # quita acentos
    s = ''.join(c for c in unicodedata.normalize('NFKD', s) if not unicodedata.combining(c))
    # minúsculas
    s = s.lower()
    # espacios y símbolos a _
    s = re.sub(r'[^a-z0-9]+', '_', s).strip('_')
    return s

# Renombrar columnas
for c in df.columns:
    df = df.withColumnRenamed(c, normalize_col(c))

df.printSchema()
df.show(5, truncate=False)

# Opcional: estandarizar nombres esperados (si existen en tu archivo)
# Intentamos mapear posibles nombres comunes:
posibles = {
    'anio':'ano', 'a_o':'ano', 'ano':'ano',
    'mes':'mes',
    'region':'region', 'regi_n':'region',
    'cultivo':'cultivo',
    'superficie_ha':'superficie_ha',
    'produccion_ton':'produccion_ton',
    'precio_ton':'precio_ton',
    'costos_insumos':'costos_insumos',
    'ingresos':'ingresos',
    'utilidad':'utilidad'
}
for viejo, nuevo in posibles.items():
    if viejo in df.columns and viejo != nuevo:
        df = df.withColumnRenamed(viejo, nuevo)

# Casteo a tipos numéricos donde aplica (si existen esas columnas)
num_cols = ["ano","mes","superficie_ha","produccion_ton","precio_ton","costos_insumos","ingresos","utilidad"]
for c in num_cols:
    if c in df.columns:
        df = df.withColumn(c, F.col(c).cast("double"))

df.printSchema()
````

3) Definir una función Python para categorizar salarios

Regla simple:

bajo: < 500.000

medio: >= 500.000 y <= 1.000.000

alto: > 1.000.000

````python
def categorizar_salario(s):
    if s is None:
        return "desconocido"
    if s < 500000:
        return "bajo"
    if s <= 1000000:
        return "medio"
    return "alto"
````

````python
from pyspark.sql.functions import udf, col
from pyspark.sql.types import StringType

# Registrar como UDF (para usar en DataFrame API)
udf_categoria = udf(categorizar_salario, StringType())

df_cat = df.withColumn("segmento_salarial", udf_categoria(col("salario_mensual")))
df_cat.show()
````

5) Registrar para SQL y aplicar en una consulta

````python
# Registrar para uso en sentencias SQL
spark.udf.register("categoria_salarial", categorizar_salario, StringType())

# Crear vista temporal
df.createOrReplaceTempView("empleados")

# Usar la UDF en SQL
resultado_sql = spark.sql("""
SELECT
  nombre,
  salario_mensual,
  categoria_salarial(salario_mensual) AS segmento_salarial
FROM empleados
ORDER BY salario_mensual DESC NULLS LAST
""")
resultado_sql.show()
````

## Desarrollo completo del ejercicio (ejecutable localmente)

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import explode, split, current_timestamp, window

# Crear SparkSession con soporte para Structured Streaming
spark = SparkSession.builder \
    .appName("StructuredStreamingVentana") \
    .getOrCreate()

# Leer datos desde socket TCP (requiere que se esté ejecutando 'nc -lk 9999' en otro terminal)
lines = spark.readStream \
    .format("socket") \
    .option("host", "localhost") \
    .option("port", 9999) \
    .load()

# Agregar timestamp actual (simulado para cada evento)
lines_with_timestamp = lines.withColumn("timestamp", current_timestamp())

# Explode y separar palabras
words = lines_with_timestamp.select(
    explode(split(lines_with_timestamp.value, " ")).alias("word"),
    "timestamp"
)

# Aplicar watermark y ventana de agregación
windowed_counts = words \
    .withWatermark("timestamp", "5 minutes") \
    .groupBy(
        window(words.timestamp, "10 minutes", "5 minutes"),
        words.word
    ).count()

# Escribir resultados en consola
query = windowed_counts.writeStream \
    .outputMode("update") \
    .format("console") \
    .option("truncate", False) \
    .start()

query.awaitTermination()
```

Categóricas: region, tiene_tarjeta

Objetivo: label (1/0)

````python
from pyspark.sql.types import StructType, StructField, IntegerType, DoubleType, StringType
from pyspark.sql import Row

schema = StructType([
    StructField("edad", IntegerType(), True),
    StructField("ingresos_mensuales", DoubleType(), True),
    StructField("region", StringType(), True),         # categórica
    StructField("tiene_tarjeta", StringType(), True),  # categórica (sí/no)
    StructField("label", IntegerType(), True)
])

data = [
    (28,  550000.0, "norte",        "si",  1),
    (35, 1200000.0, "centro",       "si",  1),
    (22,  300000.0, "sur",          "no",  0),
    (44,  800000.0, "metropolitana","si",  1),
    (31,  450000.0, "centro",       "no",  0),
    (52, 2000000.0, "metropolitana","si",  1),
    (27,  380000.0, "norte",        "no",  0),
    (39,  950000.0, "sur",          "si",  1),
    (46,  620000.0, "centro",       "no",  0),
    (33,  720000.0, "metropolitana","si",  1),
]

df = spark.createDataFrame(data, schema=schema)
df.show(truncate=False)
df.printSchema()
````

````python
from pyspark.ml.feature import VectorAssembler

features_cols = ["edad", "ingresos_mensuales", "region_idx", "tiene_tarjeta_idx"]
assembler = VectorAssembler(inputCols=features_cols, outputCol="features")
````

````python
from pyspark.ml.classification import LogisticRegression

lr = LogisticRegression(featuresCol="features", labelCol="label")
````

6) Pipeline (Indexers → Assembler → LR)

````python
from pyspark.ml import Pipeline

pipeline = Pipeline(stages=[idx_region, idx_tarjeta, assembler, lr])
````

````python
train_df, test_df = df.randomSplit([0.8, 0.2], seed=42)

model = pipeline.fit(train_df)
pred = model.transform(test_df)

pred.select("edad","ingresos_mensuales","region","tiene_tarjeta","label","probability","prediction").show(truncate=False)
````

3) Preparar features (VectorAssembler)

````python
from pyspark.ml.feature import VectorAssembler

feature_cols = ["distance_km", "stops", "hour_of_day", "traffic_index"]

assembler = VectorAssembler(
    inputCols=feature_cols,
    outputCol="features"
)

data = assembler.transform(df).select("features", df["time_min"].alias("label"))
data.show(5, truncate=False)
````

4) Split train/test

````python
train_df, test_df = data.randomSplit([0.7, 0.3], seed=42)
print("Train:", train_df.count(), " | Test:", test_df.count())
````

5) Modelo: Regresión lineal + Pipeline

````python
from pyspark.ml.regression import LinearRegression
from pyspark.ml import Pipeline

lr = LinearRegression(featuresCol="features", labelCol="label")

pipeline = Pipeline(stages=[lr])
model = pipeline.fit(train_df)
````

6) Métricas de evaluación (RMSE, MAE, R²)

````python
from pyspark.ml.evaluation import RegressionEvaluator

pred = model.transform(test_df)
pred.select("label","prediction").show(10)

e_rmse = RegressionEvaluator(labelCol="label", predictionCol="prediction", metricName="rmse")
e_mae  = RegressionEvaluator(labelCol="label", predictionCol="prediction", metricName="mae")
e_r2   = RegressionEvaluator(labelCol="label", predictionCol="prediction", metricName="r2")

rmse = e_rmse.evaluate(pred)
mae  = e_mae.evaluate(pred)
r2   = e_r2.evaluate(pred)

print(f"RMSE: {rmse:.2f}")
print(f"MAE : {mae:.2f}")
print(f"R²  : {r2:.2f}")
````

Instalar e iniciar Spark

````python
!pip install -q pyspark

from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("KMeansClientes")
    .getOrCreate()
)

spark
````

2) Generar dataset simulado (edad, ingresos, frecuencia)

````python
from pyspark.sql import Row

data = [
    (25, 1500, 5),
    (32, 2500, 10),
    (40, 4000, 15),
    (50, 5000, 8),
    (28, 1800, 6),
    (36, 3000, 12),
    (55, 7000, 20),
    (48, 4500, 9),
    (23, 1200, 3),
    (60, 8000, 25),
    (29, 2000, 7),
    (33, 2700, 11),
    (46, 4800, 14),
    (52, 6000, 18),
    (39, 3500, 13)
]

df = spark.createDataFrame(data, ["edad", "ingresos", "frecuencia"])
df.show()
````

3) Ensamblar features

````python
from pyspark.ml.feature import VectorAssembler

assembler = VectorAssembler(
    inputCols=["edad", "ingresos", "frecuencia"],
    outputCol="features"
)
df_features = assembler.transform(df)
df_features.select("edad","ingresos","frecuencia","features").show(5, truncate=False)
````

4) K-Means (k=4) + entrenamiento + predicciones

Usamos predictionCol="prediction" (el nombre por defecto) para que el evaluador funcione sin errores.

````python
from pyspark.ml.clustering import KMeans

kmeans = (
    KMeans()
    .setK(4)
    .setSeed(42)
    .setFeaturesCol("features")
    .setPredictionCol("prediction")  # importante para el Evaluator
)

model = kmeans.fit(df_features)
predicciones = model.transform(df_features)

predicciones.select("edad","ingresos","frecuencia","prediction").show()
````

5) Evaluación con Silhouette Score

````python
from pyspark.ml.evaluation import ClusteringEvaluator

evaluator = ClusteringEvaluator(
    featuresCol="features",
    predictionCol="prediction",     # coincide con KMeans
    metricName="silhouette",
    distanceMeasure="squaredEuclidean"
)

silhouette = evaluator.evaluate(predicciones)
print(f"Silhouette Score: {silhouette:.4f}")
````

6) Centros de los clústeres

````python
centros = model.clusterCenters()
print("\nCentros de los clústeres (edad, ingresos, frecuencia):")
for i, c in enumerate(centros):
    print(f"Cluster {i}: {c}")
````

7) (Opcional) Visualización 2D con PCA

Proyectar a 2 componentes principales y graficamos en 2D coloreando por prediction.

````python
from pyspark.ml.feature import PCA
import matplotlib.pyplot as plt

# PCA a 2D
pca = PCA(k=2, inputCol="features", outputCol="pca2")
pca_model = pca.fit(df_features)
df_pca = pca_model.transform(df_features)
df_pca_pred = model.transform(df_pca).select("pca2","prediction")

# Pasar a Pandas para graficar
pdf = df_pca_pred.toPandas()
xs = [vec[0] for vec in pdf["pca2"]]
ys = [vec[1] for vec in pdf["pca2"]]
clusters = pdf["prediction"].values

plt.figure(figsize=(6,5))
scatter = plt.scatter(xs, ys, c=clusters, alpha=0.8)
plt.xlabel("PC1"); plt.ylabel("PC2")
plt.title("K-Means – Proyección PCA (2D)")
plt.colorbar(scatter, label="Cluster")
plt.tight_layout()
plt.show()
````
