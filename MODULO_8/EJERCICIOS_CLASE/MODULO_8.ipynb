## Solución Propuesta:
Diagnóstico y Solución ante un Problema de Integración de Datos

### Diagnóstico del Problema

La organización presenta **bases de datos aisladas** entre las áreas de ventas, marketing y logística. Este silo de información genera:

* **Retrasos en la generación de reportes**.
* **Errores por duplicación o inconsistencia de datos**.
* **Toma de decisiones poco confiable o desactualizada**.

Este escenario es típico de una **falta de integración de datos**, donde cada área opera con su propia fuente sin una sincronización centralizada, lo que impide una visión única del negocio.

###  Riesgos operativos y estratégicos

* **Falta de trazabilidad** y duplicación de registros.
* **Reportes manuales propensos a errores**.
* **Desalineación entre equipos** por trabajar con distintos criterios o cifras.
* **Pérdida de oportunidades comerciales** al no contar con datos actualizados para la toma de decisiones.

---

###  Tipo de Integración Recomendada

Proponemos una **integración en modo *batch* (por lotes)** como primera etapa, con posibilidad de evolucionar hacia un esquema híbrido con procesos en **tiempo real** para los casos críticos.

* **Batch** es adecuada por el volumen de datos y la posibilidad de procesar de forma programada durante horas de menor carga.
* Permite integrar fuentes heterogéneas (Excel, bases relacionales, archivos planos) de forma controlada.
* Posteriormente, procesos críticos (como stock en tiempo real o alertas de clientes) podrían complementarse con integración en **tiempo real** mediante herramientas como **Apache Kafka** o **Apache Nifi**.

---

### Tareas del Proceso de Integración (ETL)

1. **Extracción (E):**

   * Lectura de datos desde distintas fuentes: hojas de cálculo, CRM, ERP, bases SQL, etc.

2. **Transformación (T):**

   * Normalización de formatos (fechas, monedas, campos comunes).
   * Limpieza de duplicados, nulos o registros inconsistentes.
   * Enriquecimiento de datos cruzando fuentes (por ejemplo, asociar datos de clientes entre ventas y marketing).

3. **Carga (L):**

   * Carga periódica a un repositorio central: puede ser un **Data Warehouse** o una **Base de Datos Integrada Operacional (ODS)**.

---

### Validaciones de Calidad de Datos

* **Chequeo de duplicados** entre sistemas.
* **Verificación de integridad referencial** (por ejemplo, que los códigos de producto existan en todas las áreas).
* **Consistencia entre fuentes** (por ejemplo, que los montos de ventas coincidan entre sistemas).
* Uso de reglas de validación y perfilado de datos antes de integrarlos.

---

### Herramientas Sugeridas

* **Pentaho Data Integration (PDI):** herramienta open source robusta para procesos ETL.
* **Talend Open Studio:** permite orquestar procesos complejos y tiene conectores para múltiples fuentes.
* **Apache Nifi:** útil si se quiere escalar hacia flujos de datos en tiempo real.
* **Airbyte** o **Fivetran**: si se buscan conectores rápidos en la nube.

---

### Propuesta Final

> La organización enfrenta un problema clásico de **fragmentación de datos** entre áreas operativas. Recomendamos implementar una solución de **integración de datos en modo batch**, utilizando herramientas como **Pentaho** o **Talend**, con procesos ETL que permitan unificar, limpiar y consolidar los datos en un solo repositorio.
>
> Esto permitirá generar reportes consistentes, reducir errores manuales y dar una visión única del cliente y de las operaciones. Como siguiente paso, se puede escalar a una integración híbrida que contemple **casos críticos en tiempo real**, para mejorar la capacidad de reacción del negocio.

---

### Beneficios Esperados

* **Visión única del negocio (single source of truth).**
* **Reportes más ágiles y confiables.**
* **Reducción de tareas manuales y errores.**
* **Toma de decisiones basada en datos unificados y actualizados.**
* **Mayor coordinación interáreas.**

#  Ejercicio: Diseño de flujo de ingesta batch para consolidación de ventas regionales

##  1. Objetivo del flujo de ingesta

El objetivo principal de este flujo es **consolidar diariamente la información de ventas provenientes de múltiples sucursales**, cuyas fuentes son archivos `.csv` heterogéneos. Esta consolidación busca:

- Unificar la estructura de datos para análisis centralizado.
- Reducir errores y duplicaciones generadas por formatos variables.
- Responder a la necesidad del área de Business Intelligence de contar con datos **listos para el análisis diario** en un único repositorio confiable y limpio.

---

##  2. Estructura del flujo ETL

###  Extracción

- **Origen de datos**: Archivos `.csv` exportados desde los sistemas de ventas de cada sucursal.
- **Ubicación**: Carpeta compartida por FTP/SFTP, bucket en cloud (por ejemplo, AWS S3), o rutas específicas en servidores internos.
- **Frecuencia**: Diaria (ejecutado cada noche fuera del horario de operaciones).
- **Nombre de archivos**: Podrían variar por región y fecha (`ventas_sucursal01_20250915.csv`, etc.).

###  Transformación

- **Limpieza**:
  - Eliminación de filas vacías o duplicadas.
  - Manejo de caracteres especiales y codificaciones.
- **Conversión de formatos**:
  - Uniformar columnas como fecha (a formato ISO `YYYY-MM-DD`), moneda, y nombres de campos (`producto_id` vs `id_producto`).
- **Estandarización**:
  - Renombrar columnas y unificar tipos de datos.
  - Aplicar reglas de negocio: eliminar productos fuera de catálogo, transformar códigos de sucursal a nombres estándar.

###  Carga

- **Destino final**:
  -  Base de datos relacional central (por ejemplo, PostgreSQL o MySQL).
  -  O bien, un **Data Warehouse** (como Snowflake, BigQuery o Amazon Redshift) para análisis en herramientas como Power BI o Tableau.
- **Estructura destino**: Tabla única consolidada `ventas_consolidadas` con campos normalizados.

---

##  3. Herramienta ETL seleccionada

###  Herramienta: **Apache NiFi**

**Justificación:**

- Permite **procesos batch agendados** y altamente configurables.
- Cuenta con **nodos visuales** para lectura, limpieza, transformación y carga.
- Escalable y extensible con componentes personalizados.
- Monitoreo de flujos en tiempo real, reinicios automáticos en caso de error.

 Alternativas válidas: **Talend Open Studio** (fácil de usar), **Pentaho Data Integration (PDI)** (interfaz visual robusta).

---

##  4. Buenas prácticas incluidas

1. **Validación de datos entrantes**:
   - Verificación de estructura y tipos de datos en cada archivo antes de procesarlo.
   - Archivos que no cumplen son enviados a una carpeta de cuarentena (`/errores/`).

2. **Trazabilidad y registros (logging)**:
   - Cada etapa (extracción, transformación, carga) registra logs en tiempo real.
   - Se mantiene historial de ejecuciones exitosas y fallidas (auditoría).

3. **Ejecución programada fuera del horario de carga**:
   - Se programa entre las 23:00 y 02:00, cuando hay baja actividad en los sistemas.

4. **Manejo de errores y notificaciones**:
   - Si falla un nodo del flujo, el sistema reintenta y, si falla de nuevo, notifica al equipo por email o Slack.

---

##  5. Representación del flujo (esquema paso a paso)

```text
 Entrada diaria (.csv desde sucursales)
        │
        ▼
 Extracción (Apache NiFi: GetFile / ListSFTP / FetchS3Object)
        │
        ▼
 Transformación (NiFi Processors: ReplaceText, UpdateAttribute, ConvertRecord)
        ├─ Validación de estructura y tipo de datos
        ├─ Normalización de columnas (fechas, montos)
        └─ Eliminación de duplicados y registros inválidos
        │
        ▼
 Carga (PutDatabaseRecord → PostgreSQL / Snowflake / BigQuery)
        │
        ▼
 Datos disponibles para BI (power BI, Tableau, Metabase)

```

El diseño del flujo batch propuesto permite consolidar diariamente los datos de ventas de múltiples sucursales, garantizando su calidad, estandarización y disponibilidad para análisis. Al utilizar herramientas como Apache NiFi y aplicar buenas prácticas, se obtiene un sistema robusto, trazable y escalable.

#  Ejercicio: Diseño de flujo ETL batch para consolidación de clientes multi-origen

##  1. Objetivos del flujo

###  Problema que resuelve
Actualmente, la empresa de retail mantiene registros de clientes dispersos en distintos sistemas (tiendas físicas, sitio web y app móvil), con problemas de:
- **Fragmentación**: los datos están distribuidos en múltiples orígenes sin integración.
- **Duplicación**: un mismo cliente puede estar registrado varias veces con datos inconsistentes.
- **Falta de estandarización**: diferentes formatos, campos y estructuras dificultan el análisis.

###  Datos a consolidar
- Información básica de clientes: nombres, apellidos, correo electrónico, teléfono, dirección.
- Datos de comportamiento: frecuencia de compra, canal de adquisición, última interacción.
- Identificadores únicos (cuando existen): RUT, email, ID de usuario app/web.

El objetivo final es construir una **base de clientes única, unificada y confiable**, que se actualice de forma **diaria**, para análisis de fidelización, segmentación y campañas personalizadas.

---

##  2. Proceso ETL

###  **Extracción**
- **Orígenes**:
  - Tiendas físicas: archivos `.csv` exportados diariamente desde sistemas locales.
  - Sitio web: base de datos relacional (por ejemplo, MySQL o PostgreSQL).
  - Aplicación móvil: API REST o base de datos en la nube (ej. Firebase).
- **Frecuencia**: diaria (ejecutado fuera del horario comercial).
- **Formato de entrada**: `.csv`, `.json`, `.parquet`, o conexión directa a base de datos.

---

###  **Transformación**
- **Limpieza**:
  - Eliminar registros vacíos o inválidos (sin email o RUT).
  - Corregir errores ortográficos o formatos en nombres/apellidos.
- **Normalización**:
  - Homogeneizar nombres de campos (`nombre_cliente`, `cliente_nombre`, etc.).
  - Unificar formatos de fecha, género, comuna, correo (todo en minúscula, sin espacios).
- **Deduplicación**:
  - Aplicar lógica de comparación fuzzy (por ejemplo, con el campo correo o RUT).
  - Priorizar registros más recientes o más completos.
- **Creación de ID único**:
  - Si no existe, generar un ID compuesto por correo + canal de origen.

---

###  **Carga**
- **Destino final**:
  - Una base de datos relacional centralizada (ej.: PostgreSQL o Snowflake).

# Automatización del Flujo de Ventas Multicanal

## Contexto

Una empresa con operaciones físicas y digitales enfrenta **cuellos de botella y errores** debido a la integración **manual** de grandes volúmenes de datos provenientes de e-commerce, aplicaciones móviles, marketplaces y puntos físicos. ## 1. Componentes del Flujo de Datos

### Fuentes de datos

Los datos provienen de múltiples canales heterogéneos:

* **E-commerce**: plataformas como Shopify, WooCommerce, Magento (JSON vía API o archivos CSV).
* **Tiendas físicas (POS)**: sistemas ERP, terminales de caja, exportaciones diarias.
* **Aplicaciones móviles**: eventos capturados mediante API REST, logs o SDKs.
* **Marketplaces**: como Amazon, Mercado Libre, Falabella, con APIs externas o archivos planos.

> **Justificación**: Estas fuentes tienen estructuras, formatos y frecuencia diferentes → es vital unificarlas para asegurar calidad y comparabilidad.

---

### Procesador (ETL)

Transformaciones necesarias:

* **Conversión de formatos** (CSV, JSON, XML → Parquet o Avro).
* **Limpieza**: eliminar registros vacíos, duplicados o con errores de codificación.
* **Estandarización**: unificar campos como `order_id`, `fecha_venta`, `sku`, `canal`.
* **Enriquecimiento**: cálculo de campos como `margen`, `utilidad`, integración con catálogo de productos o regiones.
* **Validación**: verificar existencia de campos clave (ej. fecha, total, SKU).

> Estas tareas permiten que los datos resultantes sean útiles para análisis y carga al DWH sin errores de calidad.

---

###  Enrutador o Decisor (Reglas de negocio)

Aplicar decisiones automáticas basadas en atributos:

* Si `canal = "marketplace"` → aplicar lógica de comisión y retención.
* Si `total_venta > $500.000` → marcar para revisión y logging especial.
* Si `canal = físico` → derivar a tabla exclusiva por tienda.
* Si `formato = CSV` → convertir a JSON antes de cargar.

> Esto permite mayor control, adaptabilidad por canal y detección anticipada de casos especiales o erróneos.

---

### Destino de datos

*  **Data Lake (zona bronce)**: guarda datos crudos con fines de trazabilidad.
* **Data Warehouse (zona oro)**: versión integrada y curada para analítica avanzada (ej. BigQuery, Redshift, Snowflake).
* **APIs internas**: para dashboards en tiempo real, alertas, microservicios internos.

> Permite acceso seguro, confiable y eficiente según el tipo de usuario o consumidor de los datos.



####  **Data Lake (Zona Bronce): para almacenamiento crudo y trazabilidad**

**¿Por qué se usa?**

* Permite **almacenar datos en bruto** (sin transformar), tal como vienen de los canales de origen.
* Acepta **todo tipo de formatos**: CSV, JSON, XML, Parquet, logs, imágenes, etc.
* Se usa como una “**caja negra**” o copia histórica por si hay errores o se necesita auditar procesos pasados.
* Facilita el trabajo con **modelos de machine learning**, que a veces necesitan datos sin agrupar ni resumir.

**Ventajas:**

* Bajo costo por GB almacenado.
* Alta escalabilidad.
* Preserva el historial completo.
* Útil para trazabilidad y control de calidad.

> Ejemplo: Se guarda aquí todo lo que viene desde el e-commerce y las apps, sin procesar, por si se necesita verificar ventas pasadas o recalcular.



####**Data Warehouse (Zona Oro): para analítica estructurada y decisiones de negocio**

**¿Por qué se usa?**

* Los datos que llegan aquí ya pasaron validaciones y tienen calidad asegurada.

**Ventajas:**

* Alta performance para SQL y herramientas como Power BI o Tableau.
* Diseñado para toma de decisiones.
* Ideal para integrar ventas, inventario y clientes en una sola vista.

> Ejemplo: Se carga aquí la información de ventas del día, ya estandarizada, para que los gerentes puedan ver en tiempo real cómo van las ventas.
